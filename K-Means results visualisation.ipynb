```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
```

```python
cluster_df = pd.read_csv("Clustering/clustering_real_2026_03_19_sent_volvo/clustering.csv")
cluster_df = cluster_df.drop(columns=["Unnamed: 0"])
```

```python
cluster_colors = np.array([ '#0072B2',  # Blue
                            '#E69F00',  # Orange
                            '#009E73',  # Teal/Green
                            '#B33F61']) # Pink/Magenta
columns = ["dur_mean", "dur_std", "ts_mean", "ts_std"]
scatter_plot = pd.plotting.scatter_matrix(
    cluster_df[columns],
    figsize = [10,10],
    c = cluster_colors[cluster_df['cluster']],
    s = 50
)

handles = [Patch(color=color, label=f'Cluster {i}') for i, color in enumerate(cluster_colors)]
plt.legend(handles=handles, bbox_to_anchor=(1,1), title="Clusters") 

plt.gcf().savefig("Figures/clustering_scatter_matrix.png")
```

```python
plt.figure(figsize=(6, 6))
plt.scatter(
    cluster_df['dur_std'],
    cluster_df['ts_std'],
    c=cluster_colors[cluster_df['cluster']],
    s=50
)

plt.xlabel('dur_std')
plt.ylabel('ts_std')
plt.title('dur_std vs ts_std')
plt.grid(False)
handles = [Patch(color=color, label=f'Cluster {i}') for i, color in enumerate(cluster_colors)]
plt.legend(handles=handles, bbox_to_anchor=(1,1), title="Clusters")
plt.savefig("Figures/clustering_dur_ts.png", bbox_inches='tight')
plt.show()
```

```python
cluster_percentages = cluster_df['cluster'].value_counts(normalize=True) * 100
print("Cluster percentages:", cluster_percentages)
print("Cluster counts: ")
print(cluster_df['cluster'].value_counts())
```
