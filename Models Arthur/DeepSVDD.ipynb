```python
import torch
from deepod.models.time_series import DeepSVDDTS

from utils.data_loader import load_volvo, load_optuna_val
from utils.evaluation import evaluate_model, cross_validate
from utils.hyperparameter_search import run_optuna_study

result_path="../Results/Volvo DeepSVDD"
```

```python
x_train, y_train, x_test, y_test, timestamps_train, timestamps_test, df_train, df_test = load_volvo(
    "../Dataset/Processed Dataset/volvoTrain.csv",
    "../Dataset/Processed Dataset/volvoTest.csv"
)
```

```python
hyperparameters = {
    "lr": ("categorical", [1e-6, 1e-5, 1e-4]),
    "epochs": ("int", 50, 200, False),
    "rep_dim": ("int", 32, 128, False),
}

df_train_optuna = load_optuna_val(df_train)

study = run_optuna_study(
    model_class=DeepSVDDTS,
    hyperparameters=hyperparameters,
    train_df=df_train_optuna,
    validation_ratio=0.20,
    n_trials=10,
    study_name="deepsvdd_optuna_study",
    storage="sqlite:///deepsvdd_optuna.db"
)

print("Best params:", study.best_params)
print("Best validation score:", study.best_value)
print("Best trial extra metrics:", study.best_trial.user_attrs)
```

```python
# Cross validation
best_params = study.best_params
static_params = {
    "seq_len": 50,
    "stride": 5,
}

params = best_params | static_params

cross_val_scores = cross_validate(train_df=df_train, model=DeepSVDDTS, params=params, result_path=result_path)
print(cross_val_scores)
```

```python
params = {
    "seq_len": 50,
    "stride": 5,
    'lr': 1e-05,
    'epochs': 187,
    'rep_dim': 57
}
model = DeepSVDDTS(**params)
model.fit(x_train)
```

```python
scores = model.decision_function(x_test)
```

# Utils results

```python
results = evaluate_model(y_test, scores, threshold=0.06217712195048079, result_path=result_path)
print(results)
```

```python
import pandas as pd
import os
df = pd.DataFrame(scores, columns=["score"])
scores_path = os.path.join(result_path, "scores.csv")
df.to_csv(scores_path, index=False)
```
